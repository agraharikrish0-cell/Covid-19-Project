```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sudalairajkumar/covid19-in-india")

print("Path to dataset files:", path)
```

```python
import os
print(os.listdir(path))
```

```python
import pandas as pd
file_path = os.path.join(path,"covid_19_india.csv")
df =pd.read_csv(file_path)
```

```python
# understand the  dataset
print(df.head())
print("Shape",df.shape)
print("Data type",df.dtypes)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df.info()
```

```python
df["Recovered"]=df["Cured"]
```

```python
df["Active"]=df["Confirmed"]-df["Deaths"]-df["Recovered"]
```

```python
df.tail(3)
```

```python
df.sample(10)
```

```python
state_data=(
    df.groupby("State/UnionTerritory")[["Confirmed","Deaths","Recovered","Active"]]
    .max()
    .reset_index()
)
```

```python
state_data["Recovered_RAtate"]=state_data["Recovered"]/state_data["Confirmed"]
state_data["Death_rate"]=state_data["Deaths"]/state_data["Confirmed"].replace(0,pd.NA)*100
```

```python
state_data.head()
```

```python
top10=(
    state_data.sort_values("Confirmed",ascending=False)
    .head(10)
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
plt.figure(figsize=(10,6))
sns.barplot(top10, x="State/UnionTerritory",y="Confirmed",palette=["Green","Pink","Blue"])
plt.title("TOP 10 Indian state/UTs by Confirmed COVID-19 Cases")
plt.xlabel("State/UT")
plt.ylabel("Confirmed Cases")
plt.tight_layout()
plt.show()
```

```python
daily_cases=(
    df.groupby("Date")["Confirmed"]
    .max()
    .reset_index()
)
sns.lineplot(data=daily_cases,x="Date",y="Confirmed",color="red")
plt.title("COVID-19 Confirmed Cases over time")
plt.xlabel("Date")
plt.ylabel("Confirmed Cases")
plt.xticks(rotation=45)
plt.show()
```
